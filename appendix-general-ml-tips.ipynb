## 如何處理過多的資料列數

- 讀入部分的資料列數：在 `pd.read_csv()` 函數中加入 `skiprows=[random_indice]`
    - <https://pandas.pydata.org/docs/reference/api/pandas.read_csv.html>
- 隨機取樣：使用 `df.sample()`
    - <https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.sample.html>
- 篩選需要的：使用布林索引 `df[boolean_indexing]`
- 刪除不需要的列數：<https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.drop.html>

## 如何處理過多的資料欄數

- 讀入部分的資料欄數：在 `pd.read_csv()` 函數中加入 `usecols=[column_names]`
    - <https://pandas.pydata.org/docs/reference/api/pandas.read_csv.html>
- 刪除遺漏值過多的欄位：使用 `Series.isnull()` 判斷。
    - <https://pandas.pydata.org/docs/reference/api/pandas.Series.isnull.html>
- 刪除變異數過小的欄位：使用 `Series.var()` 判斷。
    - <https://pandas.pydata.org/docs/reference/api/pandas.Series.var.html>

## 如何處理過多的資料欄數（續）

- 刪除與目標陣列相關係數接近 0 的欄位：使用 `Series.corr()`; `DataFrame.corr()` 判斷。
    - <https://pandas.pydata.org/docs/reference/api/pandas.Series.corr.html>
    - <https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.corr.html>
- 獲取特徵重要性：`RandomForestClassifier/Regressor`; `GradientBoostingClassifier/Regressor`
    - <https://scikit-learn.org/stable/auto_examples/ensemble/plot_forest_importances.html>
    - <https://scikit-learn.org/stable/modules/generated/sklearn.ensemble.GradientBoostingClassifier.html>
- 降維：<https://scikit-learn.org/stable/api/sklearn.decomposition.html>

## 如何處理過多的資料欄數（續）

- 選擇需要的欄位：使用華麗索引 `df[(col_1, col_2, ...)]`
- 刪除不需要的欄位：<https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.drop.html>
- 刪除遺漏值：<https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.dropna.html>

## 如何處理過多的資料欄數（續）

- 分類器正規化

|Classifier|Regularization Parameter|Note|
|:---------|:-----------------------|:---|
|`LogisticRegression`|C (Inverse strength)|Smaller C = Stronger regularization.|
|`RidgeClassifier`|alpha|Larger alpha = Stronger regularization.|
|SVC / LinearSVC|C (Inverse strength)|Smaller C = Stronger regularization.|
|SGDClassifier|alpha & penalty|Larger alpha = Stronger regularization; penalty can be set to 'l1', 'l2', or 'elasticnet'.|
|MLPClassifier|alpha|Larger alpha = Stronger regularization.|

## 如何處理類別不平衡

- Oversampling：增加少數類的樣本，SMOTE (Synthetic Minority Over-sampling Technique) `from imblearn.over_sampling import SMOTE`
    - <https://imbalanced-learn.org/stable/references/generated/imblearn.over_sampling.SMOTE.html>
- 讓模型在計算損失時，對少數類更敏感，`sklearn` 多數分類器都有參數 `class_weight='balanced'`
    - <https://scikit-learn.org/stable/modules/generated/sklearn.linear_model.LogisticRegression.html>

## 如何評估模型

- 交叉驗證：<https://scikit-learn.org/stable/modules/generated/sklearn.model_selection.cross_val_score.html>
- `sklearn.metrics`
    - <https://scikit-learn.org/stable/api/sklearn.metrics.html>
- Baseline Model
    - 回歸：用 `y` 的平均值作 $\hat{y}$ 預測。
    - 分類：用均勻分配作 $\hat{y}$ 預測、用 $y^{(train)}$ 比例作 $\hat{y}$ 預測。
